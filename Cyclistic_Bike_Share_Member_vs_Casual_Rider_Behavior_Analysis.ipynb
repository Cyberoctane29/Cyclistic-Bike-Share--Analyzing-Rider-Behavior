## **Company Background**

Cyclistic is a leading bike-share company based in Chicago, offering a network of bicycles and docking stations that provide residents and visitors with a convenient, flexible, and environmentally friendly mode of transportation. The company serves two primary customer segments: **casual riders**, who purchase single-ride or full-day passes, and **annual members**, who subscribe to yearly memberships for regular access to the bike-share service.

Cyclistic's marketing strategy has successfully built a broad customer base; however, the company believes that increasing the number of annual members is the key to achieving sustainable long-term growth and maximizing customer lifetime value.

## **Conduct a Dataset Assessment**

### Identify the Dataset Source

The **Cyclistic Rider Behavior Analysis** project uses **Cyclistic's historical bike trip data**, represented by the publicly available **Divvy trip datasets**. Although Cyclistic is a fictional company, the case study is based on real-world operational data collected from the **Divvy bike-share system** in Chicago, making the dataset an appropriate and representative source for answering the project's business question.

The dataset is obtained from the official **Divvy Trip Data Repository**, which is hosted on an **Amazon S3 storage bucket** and serves as the official distribution platform for Divvy's historical trip data. The repository maintains a publicly accessible archive of historical datasets dating back to **2013** and releases new trip datasets on a **monthly schedule** as compressed ZIP archives containing trip-level CSV files.

Within the Google Data Analytics Capstone, the datasets are referenced as being made available by **Motivate International Inc.** under the applicable licence. Under the current **Divvy Data License Agreement**, however, the underlying trip data is **owned by the City of Chicago**, while **Lyft Bikes and Scooters, LLC.**, the current operator of the Divvy bike-share system, is responsible for publishing and distributing the datasets through the official Divvy Trip Data Repository.

For this project, I selected the **twelve monthly datasets covering January 2024 through December 2024**, providing a complete calendar year of rider activity. Using an entire year of data ensures that seasonal variation, weekday and weekend travel patterns, weather-related fluctuations, and other temporal trends are represented, enabling meaningful comparisons between annual members and casual riders.

Each monthly dataset contains anonymized records of individual bike trips and includes information such as the ride identifier, bicycle type, trip start and end timestamps, station names and identifiers, geographic coordinates, and rider classification (**annual member** or **casual rider**). The Divvy dataset is also preprocessed by the data provider to remove maintenance trips performed by system staff and trips lasting less than **60 seconds**, reducing obvious operational records and likely false starts before public release.


### Understand the Dataset

The Divvy trip dataset contains historical records of individual bike-share trips made using the Cyclistic bike-share system in Chicago. Each record represents a single completed trip and captures operational information describing when the trip occurred, where it started and ended, the type of bicycle used, and whether the rider was an **annual member** or a **casual rider**.

The dataset follows a structured tabular format in which each row corresponds to one bike trip and each column represents a specific attribute describing that trip. The available variables capture temporal, spatial, operational, and rider-related information describing each trip.

While the dataset contains detailed operational trip records, it has been anonymized and does not include personally identifiable information (PII) or demographic attributes such as age, gender, income, occupation, or residential location. Consequently, the analysis focuses on observed rider behaviour rather than individual identities or demographic characteristics, ensuring compliance with privacy requirements while remaining well suited to the project's objectives.

### Examine the Dataset Structure

The Divvy trip dataset is organized as a collection of **twelve monthly CSV files**, with each file representing all recorded bike-share trips for a single calendar month. This monthly partitioning simplifies data publication and management while allowing analysts to examine individual months independently or combine them into a single annual dataset for year-long analyses.

All monthly datasets share a consistent tabular schema, allowing them to be combined into a single annual dataset without requiring structural modifications.

The dataset is stored in **CSV (Comma-Separated Values)** format, a widely supported, platform-independent file format that is compatible with relational database systems, programming languages, spreadsheet software, and business intelligence tools. This flexibility allows the same raw dataset to be incorporated into the project's SQL-based processing workflow before being used for exploratory analysis and report development.

The processed table, `cyclistic_tripdata_2024_processed`, was successfully created.

### 4.2 Populate Processed Table

An `INSERT INTO ... WITH ... SELECT` statement is used to populate the processed table by applying the validated data cleaning rules during data transfer from the raw staging table. The pipeline performs the required transformations and filtering within a single SQL operation, producing a cleaned dataset.

````python
con.sql("""
INSERT INTO cyclistic_tripdata_2024_processed (

    ride_id,
    rideable_type,
    started_at,
    ended_at,
    start_station_name,
    start_station_id,
    end_station_name,
    end_station_id,
    start_lat,
    start_lng,
    end_lat,
    end_lng,
    member_casual

)

WITH

-- -------------------------------------------------------------
-- Duplicate Records
-- -------------------------------------------------------------

deduplicated_rides AS (

    SELECT

        *,
        ROW_NUMBER() OVER (
            PARTITION BY ride_id
            ORDER BY ended_at DESC
        ) AS rn

    FROM cyclistic_tripdata_2024_raw

)

-- -------------------------------------------------------------
-- Temporal Cleaning
-- -------------------------------------------------------------

SELECT

    ride_id,
    rideable_type,
    started_at,
    ended_at,
    start_station_name,
    start_station_id,
    end_station_name,
    end_station_id,
    start_lat,
    start_lng,
    end_lat,
    end_lng,
    member_casual

FROM deduplicated_rides

WHERE rn = 1

    AND ended_at > started_at

    AND DATE_DIFF(
        'second',
        started_at,
        ended_at
    ) <= 86400;
""")
````

#### Duplicate Records

A Common Table Expression (CTE) combined with the `ROW_NUMBER()` window function is used to identify duplicate ride records while retaining a single authoritative record for each `ride_id`. Records are partitioned by `ride_id` and ordered by `ended_at` in descending order, assigning a rank to each duplicate. This ranking enables the subsequent filtering step to retain only the highest-ranked record (`rn = 1`) while excluding redundant duplicate entries from the processed dataset.

#### Missing Values

*Note: This data cleaning step was included in the original MySQL implementation but has been omitted from the DuckDB workflow. During the MySQL workflow, blank station names and station IDs were standardized to `NULL` using the `NULLIF()` function, while zero-valued destination coordinates (representing blank CSV fields imported as `0`) were also converted to `NULL`. In the DuckDB implementation, `read_csv_auto()` automatically imports empty CSV fields as `NULL` for both text and numeric columns. Consequently, these standardization steps are unnecessary because the imported data already contains the desired `NULL` representation.*

#### Temporal Cleaning

The final `SELECT` statement applies the validated temporal cleaning rules while transferring records into the processed table. Using the ranked output from the previous step, the query retains only the authoritative record for each `ride_id` (`rn = 1`) and filters out trips that do not satisfy the established temporal quality criteria. Specifically, records with negative ride durations (`ended_at <= started_at`) and trips exceeding 24 hours (`DATE_DIFF('second', started_at, ended_at) > 86400`) are excluded, ensuring that the processed dataset contains only logically valid ride durations.

The data cleaning pipeline was successfully executed by applying all validated cleaning rules during the transfer of records from the raw staging table to the processed table. #### Calculate ride_length_minutes from the exact duration between started_at and ended_at

### Task 5.3: Seasonal Analysis View

A `CASE` expression is used to derive a seasonal category from the ride start month, and the logic is encapsulated within a SQL `VIEW` (`v_cyclistic_seasonal_analysis`). Creating a view centralizes the season-mapping logic in a reusable virtual table, eliminating the need to repeat the same transformation across multiple analytical queries while preserving the original processed dataset.

````python
con.sql("""
CREATE OR REPLACE VIEW v_cyclistic_seasonal_analysis AS

SELECT

    *,

    CASE

        WHEN MONTH(started_at) IN (12, 1, 2) THEN 'Winter'
        WHEN MONTH(started_at) IN (3, 4, 5) THEN 'Spring'
        WHEN MONTH(started_at) IN (6, 7, 8) THEN 'Summer'
        ELSE 'Autumn'

    END AS season

FROM cyclistic_tripdata_2024_processed;
""")
````

The seasonal view successfully extends the processed dataset with a new `season` attribute while retaining all existing ride records and engineered features. This reusable abstraction provides a consistent foundation for season-based aggregations and comparisons, ensuring that all subsequent seasonal analyses reference the same standardized classification.

### Task 5.4: Seasonal Ride Volume

A `COUNT(*)` aggregation grouped by the engineered `season` attribute is used to summarize ride volume across the four meteorological seasons. A custom `CASE` expression in the `ORDER BY` clause preserves the natural chronological sequence of the seasons, ensuring the results are presented in calendar order rather than alphabetical order.

````python
con.sql("""
SELECT

    season,

    COUNT(*) AS total_rides

FROM v_cyclistic_seasonal_analysis

GROUP BY

    season

ORDER BY

    CASE

        WHEN season = 'Winter' THEN 1
        WHEN season = 'Spring' THEN 2
        WHEN season = 'Summer' THEN 3
        ELSE 4

    END;
""")
````

Seasonal ride volumes show a pronounced variation throughout the year. **Summer** records the highest activity with **2,212,192** rides, followed by **Autumn** with **1,770,558** rides and **Spring** with **1,323,893** rides, while **Winter** records the fewest rides at **545,438**. These results indicate that overall bike-share usage is strongly concentrated during the warmer seasons and declines substantially during winter.

*Note: **DuckDB Dialect Modification:** The original MySQL implementation utilized the `FIELD()` function to define a custom sort order for the seasonal categories. As `FIELD()` is not supported in DuckDB, it was replaced with an equivalent `CASE` expression within the `ORDER BY` clause. This modification ensures compatibility with the DuckDB SQL dialect while preserving the original query logic and analytical results.*

### Task 5.5: Seasonal Rider Type Distribution

A `COUNT(*)` aggregation grouped by `season` and `member_casual`, together with a window function (`SUM(COUNT(*)) OVER (PARTITION BY member_casual)`), is used to calculate the percentage distribution of rides across seasons within each rider type. Normalizing the seasonal distributions independently for annual members and casual riders enables seasonal riding patterns to be compared without being influenced by differences in overall ride volume between the two groups.

````python
seasonal_rider_df = con.sql("""
SELECT

    season,

    member_casual,

    COUNT(*) AS total_rides,

    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (
            PARTITION BY member_casual
        ),
        2
    ) AS seasonal_ride_percentage

FROM v_cyclistic_seasonal_analysis

GROUP BY

    season,

    member_casual

ORDER BY
    member_casual,
    CASE
    
            WHEN season = 'Winter' THEN 1
            WHEN season = 'Spring' THEN 2
            WHEN season = 'Summer' THEN 3
            ELSE 4
    
        END;
""").df()

seasonal_rider_df
````

The seasonal distributions highlight distinct differences in riding behavior between the two rider types. Casual riders exhibit a much stronger concentration of activity during **Summer**, with **43.69%** of their annual rides occurring in this season, compared with **34.39%** for annual members. In contrast, annual members maintain a more balanced distribution throughout the year and record a substantially higher proportion of rides during **Winter** (**11.76%** vs. **5.11%**). ## **Task 8: Day and Hour Interaction Analysis**

### **Introduction**

This section integrates the temporal dimensions of **day of the week** and **hour of the day** to examine riding behavior at the highest level of temporal granularity. By constructing day-hour interaction matrices, the analysis moves beyond separate daily and hourly summaries to identify how ride volume, rider composition, and ride duration vary across every hour of the week, providing a comprehensive view of temporal usage patterns.

### Task 8.1: Day-Hour Ride Volume

A `COUNT(*)` aggregation grouped by both the extracted day of the week and `hour_of_day` is used to calculate ride volume for every day-hour combination. Grouping across these two temporal dimensions creates a **7 × 24 interaction matrix**, enabling ride activity to be examined simultaneously by day and hour rather than through separate daily or hourly summaries.

````python
day_hour_df = con.sql("""
SELECT

    DAYNAME(ride_date) AS day_name,

    hour_of_day,

    COUNT(*) AS total_rides

FROM cyclistic_tripdata_2024_processed

GROUP BY

    WEEKDAY(ride_date),

    DAYNAME(ride_date),

    hour_of_day

ORDER BY

    WEEKDAY(ride_date),

    hour_of_day;
""").df()

day_hour_df.head(10)
````

````python
from matplotlib.colors import LinearSegmentedColormap

heatmap_cmap = LinearSegmentedColormap.from_list(
    "cyclistic_blue",
    ["#BDDCE7", "#3C92A6"]
)

plt.figure(figsize=(14, 5))

sns.heatmap(
    day_hour_df
        .pivot(
            index="day_name",
            columns="hour_of_day",
            values="total_rides"
        )
        .reindex([
            "Sunday",
            "Monday",
            "Tuesday",
            "Wednesday",
            "Thursday",
            "Friday",
            "Saturday"
        ]),
    cmap=heatmap_cmap,
    linewidths=0.5,
    cbar_kws={"label": "Total Rides"}
)

plt.title("Ride Volume by Day of Week and Hour of Day")
plt.xlabel("Hour of Day", labelpad=10)
plt.ylabel("Day of Week", labelpad=10)

plt.tight_layout()
plt.show()
````

The day-hour interaction reveals two distinct temporal demand patterns. **Weekdays** exhibit a **bimodal distribution**, with pronounced peaks during the morning commuting period (around **08:00**) and the late-afternoon commuting period (around **17:00**), reflecting typical work-related travel. In contrast, **weekends** display a **unimodal distribution**, where ride activity increases gradually throughout the morning, remains elevated during the afternoon, and declines into the evening without a distinct morning commute peak. These results indicate that weekday usage is primarily driven by commuting behavior, whereas weekend usage is characterized by more continuous recreational riding throughout the day.

### Task 8.2: Day-Hour Rider Distribution

A `COUNT(*)` aggregation grouped by the extracted day of the week, `hour_of_day`, and `member_casual`, together with a window function (`SUM(COUNT(*)) OVER (PARTITION BY member_casual)`), is used to calculate the percentage distribution of rides across every day-hour combination for each rider type. This approach creates a **7 × 24 interaction matrix** for both cohorts, enabling temporal riding patterns to be compared simultaneously across the week and throughout the day.

````python
day_hour_rider_df = con.sql("""
SELECT

    DAYNAME(ride_date) AS day_name,

    hour_of_day,

    member_casual,

    COUNT(*) AS total_rides,

    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (
            PARTITION BY
               member_casual
        ),
        2
    ) AS ride_percentage

FROM cyclistic_tripdata_2024_processed

GROUP BY

    WEEKDAY(ride_date),

    DAYNAME(ride_date),

    hour_of_day,

    member_casual

ORDER BY

	member_casual,
    
    WEEKDAY(ride_date),

    hour_of_day;
""").df()

day_hour_rider_df
````

The day-hour interaction reveals distinct temporal behaviors for the two rider types. Annual members exhibit strong **weekday commuting peaks**, with ride activity concentrated during the morning (**07:00–08:00**) and late afternoon (**16:00–17:00**) on working days. These results indicate that member riding behavior closely follows structured commuting schedules, whereas casual riders demonstrate more flexible temporal usage patterns associated with recreational and leisure travel.

### Task 10.3: Start Station Rider Distribution

A `Common Table Expression (CTE)` is used to first identify the **top 20** start stations based on total ride volume. The resulting subset is then joined back to the processed dataset, where a `COUNT(*)` aggregation grouped by `start_station_name` and `member_casual`, together with a window function (`SUM(COUNT(*)) OVER (PARTITION BY member_casual)`), is used to calculate the percentage distribution of rides across the highest-volume origin stations for each rider cohort. Restricting the analysis to the busiest stations enables rider composition to be compared at the network's most significant origin points.

````python
con.sql("""
WITH top_start_stations AS (

    SELECT

        start_station_name,

        COUNT(*) AS total_station_rides

    FROM cyclistic_tripdata_2024_processed

    WHERE start_station_name IS NOT NULL

    GROUP BY

        start_station_name

    ORDER BY

        total_station_rides DESC

    LIMIT 20

)

SELECT

    t.start_station_name,

    p.member_casual,

    COUNT(*) AS total_rides,

    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (
            PARTITION BY p.member_casual
        ),
        2
    ) AS ride_percentage

FROM top_start_stations t

JOIN cyclistic_tripdata_2024_processed p

    ON t.start_station_name = p.start_station_name

GROUP BY

    t.start_station_name,

    p.member_casual

ORDER BY

    p.member_casual,
    
    MAX(t.total_station_rides) DESC;
    
""")
````

The rider distributions reveal clear geographic differences between annual members and casual riders. Casual riders are highly concentrated at stations associated with major recreational and tourist destinations, with **Streeter Dr & Grand Ave** accounting for **16.25%** of all casual ride departures, followed by **DuSable Lake Shore Dr & Monroe St** (**10.84%**) and **Michigan Ave & Oak St** (**8.00%**). In contrast, annual members exhibit a more dispersed distribution across stations, with greater representation at locations such as **Wells St & Elm St** (**5.78%**), **Clinton St & Washington Blvd** (**5.73%**), and **State St & Chicago Ave** (**5.38%**). These findings indicate that casual riders are more strongly concentrated around recreational and tourist-oriented locations, whereas annual members make greater use of stations distributed throughout commercial and residential areas.

### Task 10.4: End Station Rider Distribution

A `Common Table Expression (CTE)` is used to first identify the **top 20** end stations based on total ride volume. The resulting subset is then joined back to the processed dataset, where a `COUNT(*)` aggregation grouped by `end_station_name` and `member_casual`, together with a window function (`SUM(COUNT(*)) OVER (PARTITION BY member_casual)`), is used to calculate the percentage distribution of rides across the highest-volume destination stations for each rider cohort. Restricting the analysis to the busiest stations enables rider composition to be compared at the network's most significant destination points.

````python
con.sql("""
WITH top_end_stations AS (

    SELECT

        end_station_name,

        COUNT(*) AS total_station_rides

    FROM cyclistic_tripdata_2024_processed

    WHERE end_station_name IS NOT NULL

    GROUP BY

        end_station_name

    ORDER BY

        total_station_rides DESC

    LIMIT 20

)

SELECT

    t.end_station_name,

    p.member_casual,

    COUNT(*) AS total_rides,

    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (
            PARTITION BY p.member_casual
        ),
        2
    ) AS ride_percentage

FROM top_end_stations t

JOIN cyclistic_tripdata_2024_processed p

    ON t.end_station_name = p.end_station_name

GROUP BY

    t.end_station_name,

    p.member_casual

ORDER BY

	p.member_casual,
    
    MAX(t.total_station_rides) DESC;
""")
````

The rider distributions closely resemble those observed for the most popular start stations, indicating consistent geographic usage patterns throughout the network. Casual riders remain heavily concentrated at destination stations associated with recreational and tourist locations, with **Streeter Dr & Grand Ave** accounting for **17.87%** of all casual ride arrivals, followed by **DuSable Lake Shore Dr & Monroe St** (**10.42%**) and **Michigan Ave & Oak St** (**8.42%**). Annual members continue to exhibit a more evenly distributed pattern across destination stations, with relatively higher representation at locations such as **Clinton St & Jackson Blvd** (**5.59%**), **Wells St & Elm St** (**5.55%**), and **State St & Chicago Ave** (**5.40%**). These findings indicate that the geographic preferences observed at trip origins persist at trip destinations, with casual riders remaining concentrated around recreational areas while annual members make greater use of stations serving commercial and residential districts.

### Task 10.5: Most Popular Routes

A `COUNT(*)` aggregation grouped by both `start_station_name` and `end_station_name` is used to calculate the total number of rides for each unique origin–destination pair. Records with missing station names are excluded using `WHERE ... IS NOT NULL` conditions to ensure only valid station-to-station trips are included. Ordering the results by ride volume in descending order identifies the highest-traffic travel corridors across the Cyclistic network.

````python
route_df = con.sql("""
SELECT

    start_station_name,

    end_station_name,

    COUNT(*) AS total_rides

FROM cyclistic_tripdata_2024_processed

WHERE

    start_station_name IS NOT NULL

    AND end_station_name IS NOT NULL

GROUP BY

    start_station_name,

    end_station_name

ORDER BY

    total_rides DESC,

    start_station_name,

    end_station_name

LIMIT 20;
""").df()

route_df 
````

### Task 10.6: Route Rider Distribution

A `COUNT(*)` aggregation grouped by `start_station_name`, `end_station_name`, and `member_casual`, together with a Common Table Expression (CTE), an `INNER JOIN`, and a window function (`SUM(COUNT(*)) OVER (PARTITION BY member_casual)`), is used to calculate the percentage distribution of rides across the top 20 origin–destination routes for each rider cohort. Restricting the analysis to the highest-volume routes enables rider composition to be examined at the most important travel corridors while excluding less frequently used routes.

````python
con.sql("""
WITH top_routes AS (

    SELECT

        start_station_name,

        end_station_name,

        COUNT(*) AS total_route_rides

    FROM cyclistic_tripdata_2024_processed

    WHERE

        start_station_name IS NOT NULL

        AND end_station_name IS NOT NULL

    GROUP BY

        start_station_name,

        end_station_name

    ORDER BY

        total_route_rides DESC

    LIMIT 20

)

SELECT

    t.start_station_name,

    t.end_station_name,

    p.member_casual,

    COUNT(*) AS total_rides,

    ROUND(
        100.0 * COUNT(*) /
        SUM(COUNT(*)) OVER (
            PARTITION BY
                p.member_casual
        ),
        2
    ) AS ride_percentage

FROM top_routes t

JOIN cyclistic_tripdata_2024_processed p

    ON t.start_station_name = p.start_station_name

    AND t.end_station_name = p.end_station_name

GROUP BY

    t.start_station_name,

    t.end_station_name,

    p.member_casual

ORDER BY
    
	p.member_casual,
     
    MAX(t.total_route_rides) DESC,

    t.start_station_name,

    t.end_station_name;
""")
````

The route-level distributions reveal pronounced behavioral differences between the two rider types. Casual riders dominate recreational and waterfront routes, with **Streeter Dr & Grand Ave → Streeter Dr & Grand Ave** accounting for **16.68%** of rides among the top routes for casual users, followed by **DuSable Lake Shore Dr & Monroe St → DuSable Lake Shore Dr & Monroe St** (**13.62%**) and **DuSable Lake Shore Dr & Monroe St → Streeter Dr & Grand Ave** (**9.90%**). In contrast, annual members are more strongly represented on inland commuter-oriented routes, including **Loomis St & Lexington St → Morgan St & Polk St** (**6.96%**) and **University Ave & 57th St → Kimbark Ave & 53rd St** (**5.01%**). These results indicate that casual riders are concentrated on recreational travel corridors and round-trip leisure routes, whereas annual members primarily utilize routes associated with regular commuting and localized point-to-point travel.

### **Business Conclusions**

#### Overall Conclusion

The analysis concludes that **annual members and casual riders exhibit fundamentally different usage patterns**, reflecting distinct transportation needs and motivations. Annual members primarily use Cyclistic for routine transportation, characterized by shorter ride durations, consistent weekday usage, and pronounced peak-hour activity. In contrast, casual riders demonstrate longer, more leisure-oriented trips, stronger weekend activity, and increased ridership during warmer months.

These behavioural differences indicate that **casual riders represent the greatest opportunity for annual membership growth**. Rather than attempting to change how casual riders use the service, Cyclistic should align its membership strategy with their existing recreational behaviour by offering value propositions and membership options that appeal to highly engaged casual riders. This provides a targeted path toward increasing annual memberships while building on established rider behaviour.

#### Highest-Impact Findings

Three findings have the greatest business impact because they directly inform Cyclistic's membership-growth strategy:

1. **Casual riders demonstrate longer, leisure-oriented usage.**  
   Casual riders have a higher average ride duration and are more strongly represented across longer-duration categories, indicating greater recreational use of the service. The distribution analysis shows that **70.80% of casual rides are under 20 minutes**, compared with **85.43% of member rides**, while casual riders have substantially higher proportions of rides in the 20–30 and 30–60 minute ranges.

2. **Annual members demonstrate predictable commuter-oriented usage.**  
   Members show a stronger pattern of shorter, recurring rides during weekdays and peak commuting periods, indicating that the existing membership model aligns particularly well with routine transportation needs. This establishes a clear behavioural contrast with the more recreational casual-rider segment.

3. **Casual rider engagement increases during weekends and warmer months.**  
   Casual activity rises during periods associated with higher recreational demand, creating identifiable opportunities to engage riders when their interaction with the service is strongest. These periods therefore provide strategic opportunities for targeted membership-conversion campaigns.

**Overall business implication:** The strongest growth opportunity is to **convert high-engagement casual riders into annual members by aligning membership value, promotional messaging, and campaign timing with their existing recreational riding behaviour**, rather than attempting to shift them toward the commuter-oriented behaviour of current members.

### **Business Recommendations**

#### **1. Promote Leisure-Oriented Membership Value**

Cyclistic should position annual memberships as a valuable option for riders who frequently use the service for recreational and longer-duration trips. Marketing initiatives can emphasize **scenic routes, weekend riding, extended ride benefits, and leisure-focused membership messaging** to demonstrate how membership value extends beyond routine transportation.

#### **2. Reduce Commitment Barriers**

Cyclistic should introduce more flexible pathways for casual riders who may not be ready to commit to an annual membership. Potential approaches include **weekend-only membership options, seasonal or monthly memberships, short-term trials, and flexible ride bundles** that allow casual riders to experience membership benefits before making a longer-term commitment.

#### **3. Target Peak Conversion Opportunities**

Membership campaigns should be concentrated around periods and locations where casual rider engagement is highest. **Summer campaigns, weekend promotions, afternoon and leisure-focused offers, and targeted marketing at high-traffic stations** can help reach casual riders when their interaction with the service is strongest and create more relevant opportunities for membership conversion.

### **Action Planning and Measurement**

#### Implementation Priorities

Cyclistic should prioritize initiatives that target **high-engagement casual riders** during periods of strong recreational demand. Initial implementation should focus on leisure-oriented membership messaging, flexible or lower-commitment membership options, and targeted campaigns during **weekends and warmer riding seasons**. These initiatives should be introduced as targeted marketing strategies and refined based on observed performance.

#### Stakeholder Responsibilities

The **marketing team** should develop and implement targeted membership campaigns, promotional offers, and leisure-focused messaging while monitoring campaign performance. The **executive team** should provide strategic direction, allocate resources toward high-potential initiatives, support flexible membership offerings, and review overall membership-growth performance. Collaboration across **marketing, operations, and customer experience teams** will support effective implementation of the recommended strategies.

#### Success Metrics / KPIs

The effectiveness of the recommendations should be evaluated using measurable business and marketing KPIs, including:

- **Casual-to-annual membership conversion rate**
- **Annual membership growth**
- **Customer acquisition cost (CAC)**
- **Member retention rate**
- **Marketing campaign ROI**

These metrics should be monitored before and after implementation and reviewed regularly to determine which initiatives generate the strongest conversion and business impact. Continuous performance measurement will allow Cyclistic to refine its strategies and maintain alignment with its objective of increasing annual memberships.

### **Future Opportunities and Limitations**

#### Additional Data

Future analyses could be strengthened by incorporating **customer demographics**, an **anonymized customer identifier** to track individual rider behavior over time, membership purchase history, trip purpose, marketing campaign interactions, customer feedback, promotional offer redemption, weather conditions, station capacity, and special event data. These additional sources would provide deeper insight into individual rider journeys, repeat usage, rider motivations, membership decisions, and external factors influencing ridership.

#### Future Analyses

Future analysis could evaluate the **long-term effectiveness of membership conversion strategies**, compare rider behavior across multiple years, and examine how behavior changes after casual riders become annual members. Additional opportunities include more detailed customer segmentation, station-level analysis, and predictive modeling to identify casual riders with the highest likelihood of converting to annual memberships.

#### Limitations

The recommendations are based on **historical trip data** and behavioral patterns observed during the analysis period. The dataset does not include an **anonymized individual rider identifier**, limiting the ability to track individual riders across trips, analyze behavior over time, conduct **panel or cohort analyses**, and directly evaluate individual membership conversion and retention. The analysis also does not include customer demographics, marketing interactions, or qualitative feedback, limiting the ability to directly explain the motivations behind membership decisions. External factors such as weather, local events, infrastructure changes, economic conditions, and changing rider preferences may also affect future outcomes. Therefore, the recommendations should be treated as **evidence-based strategic guidance rather than guaranteed outcomes** and evaluated through continuous performance monitoring and periodic reanalysis.
